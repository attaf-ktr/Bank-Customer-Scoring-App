```python
import pandas as pd
import os


data_path = "/home/kaoutar/Desktop/data"  

# 📃 Dictionnaire des fichiers avec séparateurs
files_info = {
    "CREDIT": ("ACT_CD_CREDIT_MDM28032025.txt", "|"),
    "GETCHQ": ("ACT_CD_GETCHQ_MDM28032025.txt", "|"),
    "OPERATIONS_AGENCE": ("ACT_CD_OPERATIONS_AGENCE_MDM20032025.txt", ";"),
    "PAIEMENT_VIGNETTE": ("ACT_CD_PAI_VIGNETTE_MDM20032025.txt", "|"),
    "PRODUIT": ("ACT_CD_PRODUIT_MDM20032025.txt", "|"),
    "SOLDE_CLIENT": ("ACT_CD_SOLDE_CLIENT_MDM28032025.txt", "|"),
    "CONNEXION": ("ACT_CD_STAT_CONNEXION_MDM28032025.txt", "|"),
    "TRANSA": ("ACT_CD_TRANSA_MDM20032025.txt", "|"),
    "TRANSFERTS": ("ACT_CD_TRANSFERTS_COMPTE_MDMV20032025.txt", "|"),
    "EQUIPEMENT": ("ACT_EQUI_MDM28032025.txt", "|"),
    "PERIMETRE": ("Périmétre_projet.txt", "|")
}

# 📦 Dictionnaire pour stocker les DataFrames
dataframes = {}

# 📥 Chargement + aperçu
for name, (filename, sep) in files_info.items():
    file_path = os.path.join(data_path, filename)
    try:
        df = pd.read_csv(file_path, sep=sep, dtype=str)
        dataframes[name] = df

        print(f"\n📁 === {name} ({filename}) ===")
        print("Shape:", df.shape)
        print("Colonnes:", list(df.columns))
        print("\n🔍 Aperçu :")
        display(df.head())

        print("\n🧠 Infos générales :")
        display(df.info())

        print("\n📉 Statistiques :")
        display(df.describe(include='all'))

        print("\n❓ Valeurs manquantes :")
        display(df.isnull().sum())

    except Exception as e:
        print(f"⚠️ Erreur de chargement {filename} : {e}")
```

```python
import pandas as pd
import numpy as np

data_path = "/home/kaoutar/Desktop/data"

# Chargement
df_conn = pd.read_csv(f"{data_path}/ACT_CD_STAT_CONNEXION_MDM28032025.txt", sep="|")
df_trans = pd.read_csv(f"{data_path}/ACT_CD_TRANSA_MDM20032025.txt", sep="|")
df_getchq = pd.read_csv(f"{data_path}/ACT_CD_GETCHQ_MDM28032025.txt", sep="|")
df_vignette = pd.read_csv(f"{data_path}/ACT_CD_PAI_VIGNETTE_MDM20032025.txt", sep="|")
df_perimetre = pd.read_csv(f"{data_path}/Périmétre_projet.txt", sep="|")
```
